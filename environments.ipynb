# Código para gerar tabuleiros do experimento FrozenLake

```python
import gymnasium as gym
from gymnasium.envs.toy_text.frozen_lake import generate_random_map
import matplotlib.pyplot as plt
import pickle
```

## 8X8

```python
env8 = gym.make(
        "FrozenLake-v1",
        is_slippery=False,
        render_mode="rgb_array",
        desc=generate_random_map(
            size=8, p=0.9, seed=123
        ),
    )

env10 = gym.make(
        "FrozenLake-v1",
        is_slippery=False,
        render_mode="rgb_array",
        desc=generate_random_map(
            size=10, p=0.9, seed=123
        ),
    )

env12 = gym.make(
        "FrozenLake-v1",
        is_slippery=False,
        render_mode="rgb_array",
        desc=generate_random_map(
            size=12, p=0.9, seed=123
        ),
    )
```

```python
env8.reset()
env10.reset()
env12.reset()

map8 = env8.get_wrapper_attr("desc")
map10 = env10.get_wrapper_attr("desc")
map12 = env12.get_wrapper_attr("desc")
plt.imshow(env8.render())
plt.imshow(env10.render())
plt.imshow(env12.render())
```

```python
with open('maps/8x8_2.pkl', 'wb') as outf:
    pickle.dump(map8, outf) 
with open('maps/10x10_2.pkl', 'wb') as outf:
    pickle.dump(map10, outf) 
with open('maps/12x12_2.pkl', 'wb') as outf:
    pickle.dump(map12, outf) 
```

```python
with open('maps/8x8.pkl', 'rb') as inf: 
    loaded_array = pickle.load(inf) 
```

```python
env2=gym.make('FrozenLake-v1', desc=loaded_array, is_slippery=False, render_mode="rgb_array")
```

```python
env2.reset()
plt.imshow(env2.render())
```
